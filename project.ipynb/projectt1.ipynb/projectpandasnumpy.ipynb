```python
import numpy as np
import pandas as pd

np.random.seed(11)
df1 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])
df1['Product_AB'] = df1['A'] * df1['B']
print(df1)
```

```python
np.random.seed(22)
df2 = pd.DataFrame(np.random.randint(1, 15, size=(4, 3)), columns=['X', 'Y', 'Z'])
col_sum = df2.sum(axis=0)
row_sum = df2.sum(axis=1)

print("Original DataFrame:\n", df2)
print("\nColumn-wise Sum:\n", col_sum)
print("\nRow-wise Sum:\n", row_sum)
```

```python
np.random.seed(33)
df3 = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)).astype(float), columns=['Val1', 'Val2', 'Val3'])
df3.iloc[0, 1] = np.nan
df3.iloc[3, 2] = np.nan

df3_imputed = df3.fillna(df3.mean())
print(df3_imputed)
```

```python
# Task 4
np.random.seed(44)
df4 = pd.DataFrame(np.random.randint(1, 25, size=(6, 4)).astype(float), columns=['ColA', 'ColB', 'ColC', 'ColD'])
df4.iloc[1, 0] = np.nan
df4.iloc[4, 3] = np.nan

df4_cleaned = df4.dropna()
print(df4_cleaned)
```

```python
# Task 5
np.random.seed(55)
df5 = pd.DataFrame({
    'Category': np.random.choice(['GroupA', 'GroupB', 'GroupC'], size=10),
    'Value': np.random.randint(10, 100, size=10)
})

grouped5 = df5.groupby('Category')['Value'].agg(['sum', 'mean'])
print(grouped5)
```

```python
# Task 6
np.random.seed(66)
df6 = pd.DataFrame({
    'Product': ['Item1', 'Item2', 'Item3', 'Item4', 'Item5', 'Item6'],
    'Category': ['Electronics', 'Fashion', 'Electronics', 'Home', 'Fashion', 'Home'],
    'Sales': np.random.randint(100, 1000, size=6)
})

total_sales = df6.groupby('Category')['Sales'].sum().reset_index()
print(total_sales)
```

```python
# Task 7
df7_a = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

df7_b = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Department': ['HR', 'IT', 'Finance', 'Marketing']
})

merged_df = pd.merge(df7_a, df7_b, on='Employee_ID')
print(merged_df)
```

```python
# Task 8
index8 = pd.MultiIndex.from_tuples([
    ('Class_A', 'Student1'), ('Class_A', 'Student2'),
    ('Class_B', 'Student1'), ('Class_B', 'Student2')
], names=['Class', 'Student'])

df8 = pd.DataFrame({'Score1': [85, 90, 78, 88], 'Score2': [88, 92, 80, 85]}, index=index8)

print("Slicing Class_A:\n", df8.loc['Class_A'])
```

```python
# Task 9
index9 = pd.MultiIndex.from_product([['Region1', 'Region2'], ['TypeA', 'TypeB']], names=['Category', 'SubCategory'])
df9 = pd.DataFrame({'Units_Sold': [50, 120, 80, 150]}, index=index9)

sum_result = df9.groupby(level=['Category', 'SubCategory']).sum()
print(sum_result)
```

```python
# Task 10
np.random.seed(100)
df10 = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)), columns=['P', 'Q', 'R'])

doubled_df = df10.apply(lambda val: val * 2)
print(doubled_df)
```

```python
# Task 11
np.random.seed(111)
df11 = pd.DataFrame(np.random.randint(1, 15, size=(6, 3)), columns=['Num1', 'Num2', 'Num3'])

df11['Row_Sum'] = df11.apply(lambda r: r['Num1'] + r['Num2'] + r['Num3'], axis=1)
print(df11)
```
